# 01 · Demo 1 — the agent that *looks* fine

A toy credit-assistant agent runs four planned steps:

```
retrieve_docs → call_pricing_api → apply_discount → confirm_order
```

**The point of this demo:** the buggy run's *customer-facing output* looks perfectly fine — it even says "Order confirmed!" — but the **trace** shows it went off-task and skipped a step. Output-only testing misses this; trace-level evaluation catches it.

The agent is deterministic (no live LLM), and its trace is logged to MLflow.

````python
import os
from pathlib import Path

# Notebooks live in <repo>/notebooks; walk up to the dir that holds the Makefile.
root = Path.cwd()
while not (root / "Makefile").exists() and root != root.parent:
    root = root.parent
os.chdir(root)
print("working dir:", Path.cwd())
````

## The buggy run — off-task at step 3, step 4 skipped
Watch step 3 pitch a rewards card instead of applying the discount, step 4 never run — and the customer still told *"Order confirmed!"*

````python
!make agent-buggy
````

### What just happened
- **step 3 `apply_discount` → OFF TASK:** pitched a Platinum Rewards card; the loyalty discount was silently never applied.
- **step 4 `confirm_order` → SKIPPED:** never executed.
- **customer sees:** *"Order confirmed!"* — the output is reassuring and wrong.
